#### 1.Area Calculator: Write a function area_rect(length, width) that calculates and prints the area.

```python
def area_rect(length,width):
    print(length*width)
area_rect(5,7)
```

#### 2.Voting Eligibility: Write a function can_vote(age) that prints "Eligible" if age is 18 or older, and "Not Eligible" otherwise.

```python
def can_vote(age):
    if age>=18:
        print("Eligible")
    else:
        print("Not Eligible")
can_vote(23)        
```

#### 3.Max Number: Write a function find_max(a, b) that prints the larger of the two numbers.

#### 4.String Length: Create a function check_len(s) that prints whether a string is "Long" (more than 10 chars) or "Short".

```python
def check_len(s):
    if len(s) >= 12:
        print("Long")
    else:
        print("short")
check_len("dnyaneshwari")        
```

#### 5.Temperature Converter: Write a function that takes Celsius and prints the Fahrenheit equivalent.

```python
def temperature(celsius):
    fahrenheit = (celsius * 9/5) + 32
    print(fahrenheit)

temperature(25)
```
