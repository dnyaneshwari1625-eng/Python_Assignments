#### 6. Find Prime Numbers: Print all prime numbers between 1 and a given number N. 

```python
n = 10

for i in range(2, n + 1):
    prime = True

    for j in range(2, i):
        if i % j == 0:
            prime = False
            break

    if prime:
        print(i)
```

#### 8. Sum of Odd and Even Numbers: Take a list of numbers and find the sum of odd and even numbers separately.

```python
numbers = [1, 2, 3, 4, 5]

even_sum = 0
odd_sum = 0

for i in numbers:
    if i % 2 == 0:
        even_sum += i
    else:
        odd_sum += i

print("Sum of Even Numbers:", even_sum)
print("Sum of Odd Numbers:", odd_sum)
```

#### 3. Simple Sum Calculator: Keep asking the user for numbers and calculate the sum until they enter 0.

```python
numbers = [5, 10, 3, 0]

sum = 0
i = 0

while numbers[i] != 0:
    sum = sum + numbers[i]
    i = i + 1

print("Total Sum:", sum)
```

### Intermediate Level 

#### 4. Reverse Digits: Input a number and reverse its digits using a while loop. 

```python
num = 1234

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number:", reverse)
```

#### 5. Find the Smallest Divisor: Write a program to find the smallest divisor of a given number other than 1.

```python
num = 15

i = 2

while i <= num:
    if num % i == 0:
        print("Smallest Divisor:", i)
        break
    i = i + 1
```

### Advanced Level

#### 6. Fibonacci Sequence: Generate the first N terms of the Fibonacci sequence using a while loop.

```python
n = 5

a = 0
b = 1
count = 0

while count < n:
    print(a)
    c = a + b
    a = b
    b = c
    count = count + 1
```

#### 7. Check Palindrome Number: Write a program to check whether a number is a palindrome (e.g., 121 is a palindrome).

```python
num = 121

temp = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if temp == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
