#### 3.Vowel Counter: Create a function count_vowels(word) that loops through a string and prints the number of vowels found.

```python
def count_vowels(word):
    count = 0
    for letter in word:
        if letter.lower() in "aeiou":
            count += 1
    print(count)
```

```python
count_vowels("Hello World")
```

#### 4.Countdown: Write a function that takes a number x and uses a while loop to count down to zero.

```python
def countdown(x):
    while x >= 0:
        print(x)
        x -= 1
```

```python
countdown(5)
```

#### 5.Filtering: Write a function that takes a list of numbers and prints only the ones that are divisible by 3.

```python
def divisible_by_3(numbers):
    for n in numbers:
        if n % 3 == 0:
            print(n)
```

```python
divisible_by_3([1, 3, 6, 8, 9, 12])
```

#### 2.Simple Calculator: Create a function calc(a, b, operation) where operation can be "add", "sub", "mul", or "div".

```python
def calc(a, b, operation):
    if operation == "add":
        print(a + b)
    elif operation == "sub":
        print(a - b)
    elif operation == "mul":
        print(a * b)
    elif operation == "div":
        print(a / b)
```

```python
calc(10, 5, "add")
calc(10, 5, "sub")
calc(10, 5, "mul")
calc(10, 5, "div")
```

#### 3.Password Validator: Write a function that checks if a password is longer than 8 characters AND contains the symbol "@".

```python
def check_password(password):
    if len(password) > 8 and "@" in password:
        print("Valid")
    else:
        print("Invalid")
```

```python
check_password("hello@123")
check_password("hello123")
```

#### 4.Nested Logic: Write a function that takes a list of ages and uses the even_odd logic from your notebook to count how many even-numbered ages exist.

```python
def count_even_ages(ages):
    count = 0
    for age in ages:
        if age % 2 == 0:
            count += 1
    print(count)
```

```python
count_even_ages([12, 15, 18, 21, 24])
```
