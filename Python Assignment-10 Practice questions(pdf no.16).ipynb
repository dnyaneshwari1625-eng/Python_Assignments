### Part A: Basic List Comprehensions (Level 1)

#### 1.Squares: Create a list of squares for numbers 1 to 10.

```python
squares = [num**2 for num in range(1,11)]
print(squares)
```

#### 2.Upper Case: Given names = ['om', 'sai', 'ram'], convert them all to uppercase in one line.

```python
names = ['om', 'sai', 'ram']
upper_names = [name.upper() for name in names]
print (upper_names)
```

### Part B: Conditionals in Comprehension (Level 2)

#### 1.Even/Odd Labels: Create a list that says "Even" if the number is even and "Odd" if it is odd for range(1, 6).

```python
labels = []

for num in range(1,6):
    if num % 2 == 0:
        labels.append("Even")
    else:
        labels.append("Odd")
print(labels)        
```

#### 2.Positive Only: From data = [1, -5, 2, -8, 10], extract only the positive numbers.

```python
data = [1, -5, 2, -8, 10]

positive = []
for num in data:
    if num > 0:
        positive.append(num)
```

```python
print(positive)
```

#### 3.Vowel Remover: Given a string, create a list of all its characters except the vowels.

```python
text = 'education'

result = []

for ch in text:
    if ch not in 'aeiouAEIOU':
        result.append(ch)
```

```python
print(result)
```

#### 4.Type Filter: Given l = [1, 'a', 2.5, 'b', 3], extract only the integers.

```python
l = [1, 'a', 2.5, 'b', 3]

integers = []

for item in l:
    if type(item) == int:
        integers.append(item)
```

```python
print(integers)
```

#### 5.Short Word Filter: From a list of fruits, keep only the ones with more than 5 letters.



```python
fruits = ['apple','banana','custard','pear']

long_fruits = []

for fruit in fruits:
    if len(fruit) > 5:
        long_fruits.append(fruit)
```

```python
print(long_fruits)
```

### 6.Part C: Dictionary Comprehensions (Level 3)

#### 1.Number Mapping: Create a dictionary where the key is a number (1-5) and the value is its cube.



```python
cubes = {}

for num in range(1,6):
    cubes[num] = num**3
```

```python
print(cubes)
```

#### 2.Name Lengths: Create a dictionary where keys are names and values are their lengths.

```python
names = ['Diya','Om','Sai','Parag']

lengths = {}

for name in names:
    lengths[name] = len(name)
```

```python
print(lengths)
```

#### 3.Price Conversion: Given prices_in_usd = {'apple': 1.0, 'bread': 2.5}, create a new dictionary with prices doubled.

```python
prices_in_usd = {'apple': 1.0, 'bread': 2.5}

new_prices = {}

for item in prices_in_usd:
    new_prices[item] = prices_in_usd[item] * 2
```

```python
print(new_prices)
```

#### 4.Character Count: Create a dictionary counting occurrences of each character in "ALPHABET" (use .count()).

```python
text = 'ALPHABET'

counts = {}

for ch in text:
    counts[ch] = text.count(ch)
    
```

```python
print(counts)
```

#### 5.Conditional Dict: Create a dictionary of numbers 1-10, but only include those whose square is greater than 50.

```python
numbers = {}

for num in range(1,11):
    if num**2 > 50:
        numbers[num] = num**2
```

```python
print(numbers)
```

### Part D: Expert "Top 1%" Patterns (Level 4)


#### 1.Nested List Flattening: Convert [[1, 2], [3, 4]] into [1, 2, 3, 4] using one line of list comprehension.

```python
nested = [[1, 2], [3, 4]]

flat = []

for sublist in nested:
    for num in sublist:
        flat.append(num)
```

```python
print(flat)
```

#### 2.Matrix Creation: Create a 3x3 matrix of zeros: [[0, 0, 0] for i in range(3)].

```python
matrix = [[0, 0, 0] for i in range(3)]

print(matrix)
```

#### 3.Zip to Dict: Given keys = ['ID', 'Name'] and vals = [101, 'Python'], use zip() inside a comprehension to make a dictionary.

```python
keys = ['ID', 'Name']
vals = [101, 'Python']

data = dict(zip(keys, vals))
```
